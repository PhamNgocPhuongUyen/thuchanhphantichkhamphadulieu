# Huấn luyện mô hình bằng GRADIENT BOOSTING CLASSIFIER

```python
import pandas as pd
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import cross_val_score, GridSearchCV
import numpy as np
import joblib
# NẠP DỮ LIỆU ĐÃ TIỀN XỬ LÝ TỪ FILE preprocessing.ipynb
try:
    X_train = pd.read_csv('X_train_processed.csv')
    y_train = pd.read_csv('y_train.csv')['Survived']
    X_test = pd.read_csv('X_test_processed.csv')
    test_ids = pd.read_csv('test_ids.csv')['PassengerId']
    print("Đã nạp dữ liệu tiền xử lý thành công.")
    print(f"X_train.shape: {X_train.shape}")
except FileNotFoundError:
    print("LỖI: Không tìm thấy file dữ liệu đã tiền xử lý. Hãy chạy lại file preprocessing.ipynb.")
    import sys; sys.exit()
```

## 1. Huấn luyện và đánh giá baseline

```python
gb_base = GradientBoostingClassifier(random_state=42)
cv_scores = cross_val_score(gb_base, X_train, y_train, cv=5, scoring='accuracy')

print("\n--- Gradient Boosting Baseline ---")
print(f"Điểm CV Trung bình: {cv_scores.mean():.4f}")
```

## 2. Tối ưu hoá tham số (GRID SEARCH)

```python
# Dùng Grid Search để tìm tham số tốt nhất
param_grid = {
    'n_estimators': [100, 200],
    'learning_rate': [0.05, 0.1, 0.2],
    'max_depth': [3, 5]
}

grid_search_gb = GridSearchCV(
    estimator=GradientBoostingClassifier(random_state=42),
    param_grid=param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1
)

grid_search_gb.fit(X_train, y_train)
```

## 3. Kết quả và dự đoán cuối cùng

```python
gb_final_model = grid_search_gb.best_estimator_

print("\n--- Kết quả Tối ưu hóa Gradient Boosting ---")
print(f"Tham số tốt nhất: {grid_search_gb.best_params_}")
print(f"Điểm CV TỐT NHẤT: {grid_search_gb.best_score_:.4f}")

# Dự đoán trên tập kiểm tra
gb_predictions = gb_final_model.predict(X_test)
```

```python
submission_gb_df = pd.DataFrame({
    'PassengerId': test_ids,
    'Survived': gb_predictions.astype(int)
})

submission_gb_df.to_csv('submission_model2_GB_Final.csv', index=False) 
print("\nĐÃ LƯU: submission_model2_GB_Final.csv (Mô hình Gradient Boosting)")

joblib.dump(gb_final_model, 'gb_final_model.pkl')
print("\nĐã lưu mô hình Gradient Boosting (gb_final_model.pkl)")
```
