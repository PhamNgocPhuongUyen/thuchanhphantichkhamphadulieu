```python
import pandas as pd
from autoviz.AutoViz_Class import AutoViz_Class

# Khởi tạo AutoViz
AV = AutoViz_Class()

# Load dữ liệu
marketing_data = pd.read_csv("/content/marketing_campaign.csv", sep="\t")

# Tạo visualization tự động từ file
filename = "marketing_campaign.csv"
# depVar = biến mục tiêu nếu bạn có (nếu không để '')
dft = AV.AutoViz(
    filename,
    sep='\t',
    depVar='',             # để '' nếu không có target
    dfte=None,
    header=0,
    verbose=2,
    lowess=False,
    chart_format='svg',    # hoặc 'png'
    max_rows_analyzed=50000,
    max_cols_analyzed=50
)
# AutoViz sẽ in logs và mở/ghi hàng loạt hình ảnh / báo cáo
```

```python
# Với nhiều tùy chọn hơn
dft_advanced = AV.AutoViz(
    filename="",
    dfte=marketing_data,
    depVar="Response",
    verbose=2,  # 0: ít thông tin, 1: trung bình, 2: chi tiết
    lowess=False,  # Thêm đường xu hướng LOWESS
    chart_format="bokeh",  # "png", "svg", "bokeh", "server", "html"
    max_rows_analyzed=150000,  # Số dòng tối đa phân tích
    max_cols_analyzed=30  # Số cột tối đa phân tích
)
```
