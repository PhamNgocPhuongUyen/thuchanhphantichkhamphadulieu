```python
import pandas as pd
import sweetviz as sv
# Đọc dữ liệu
marketcp_data = pd.read_csv('/content/marketing_campaign.csv', sep='\t')
# Tạo báo cáo cho tất cả các biến
report = sv.analyze(marketcp_data)
report.show_html("marketcp_sweetviz.html")
```

```python
# Tạo báo cáo tổng quát cho biến mục tiêu "Response"
report_with_target = sv.analyze(marketcp_data, target_feat='Response')
report_with_target.show_html("Marketing_Campaign_EDA_With_Target_Report.html")
```

```python
# Tạo báo cáo so sánh 2 nhóm khách hàng: khách hàng có trẻ nhỏ và khách hàng không có trẻ nhỏ
df_has_kids = marketcp_data[marketcp_data["Kidhome"]>0]
df_no_kids = marketcp_data[marketcp_data["Kidhome"]==0]
# So sánh
report_comp = sv.compare([df_has_kids, "HasKids"], [df_no_kids, "NoKids"])
report_comp.show_html("sweetviz_compare_kids.html")
```
